## <font color="maroon"><h4 align="center">Handling Missing Data - fillna, interpolate, dropna</font>

```python
# Load pandas, read weather_data.csv with parsed dates, inspect the first row type and display the DataFrame.
import pandas as pd
df = pd.read_csv("weather_data.csv",parse_dates=['day'])
type(df.day[0])
df
```

```python
# Set the DataFrame index to the day column and display the updated DataFrame.
df.set_index('day',inplace=True)
df
```

<!-- Section header for fillna examples. -->
## <font color="blue">fillna</font>

<!-- Describe filling all missing values with a single constant. -->
<font color="purple">**Fill all NaN with one specific value**</font>

```python
# Replace all missing values with 0 and display the resulting DataFrame.
new_df = df.fillna(0)
new_df
```

<!-- Describe filling NaNs with different values for each column. -->
<font color="purple">**Fill na using column names and dict**</font>

```python
# Fill missing values using a dictionary of column-specific replacements and display the result.
new_df = df.fillna({
        'temperature': 0,
        'windspeed': 0,
        'event': 'No Event'
    })
new_df
```

<!-- Describe using method-based fill operations for missing values. -->
<font color="purple">**Use method to determine how to fill na values**</font>

```python
# Forward-fill missing values along the index and display the result.
new_df = df.ffill()
new_df
```

```python
# Backward-fill missing values along the index and display the result.
new_df = df.bfill()
new_df
```

<!-- Describe filling missing values across columns instead of rows. -->
<font color="purple">**Use of axis**</font>

```python
# Backward-fill missing values across columns and display the result.
new_df = df.bfill(axis="columns") # axis is either "index" or "columns"
new_df
```

<!-- Explain the limit parameter for fill operations. -->
<font color="purple">**limit parameter**</font>

```python
# Forward-fill missing values with a maximum of one consecutive fill and display the result.
new_df = df.ffill(limit=1)
new_df
```

<!-- Section header for interpolation examples. -->
### <font color="blue">interpolate</font>

```python
# Interpolate only the numeric temperature and windspeed columns and display the result.
new_df = df[["temperature","windspeed"]].interpolate()
new_df
```

```python
# Perform time-based interpolation on the numeric temperature and windspeed columns and display the result.
new_df = df[["temperature","windspeed"]].interpolate(method="time")
new_df
```

<!-- Note about the linear interpolation result for a specific date. -->
**Notice that in above temperature on 2017-01-04 is 29 instead of 30 (in plain linear interpolate)**

<!-- Note that pandas supports many other interpolation methods beyond linear and time-based. -->
**There are many other methods for interpolation such as quadratic, piecewise_polynomial, cubic etc. 
Just google "dataframe interpolate" to see complete documentation**

<!-- Section header for dropna examples. -->
### <font color="blue">dropna</font>

```python
# Drop any rows containing missing values and display the resulting DataFrame.
new_df = df.dropna()
new_df
```

```python
# Drop only the rows where all values are missing and display the result.
new_df = df.dropna(how='all')
new_df
```

```python
# Drop rows that do not have at least one non-missing value and display the result.
new_df = df.dropna(thresh=1)
new_df
```

<!-- Section header for inserting missing dates into the index. -->
### <font color="blue">Inserting Missing Dates</font>

```python
# Create a complete daily date range and reindex the DataFrame to include missing dates.
dt = pd.date_range("01-01-2017","01-11-2017")
idx = pd.DatetimeIndex(dt)
df.reindex(idx)
```
