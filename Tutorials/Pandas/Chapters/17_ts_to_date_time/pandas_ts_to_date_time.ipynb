<h1 style="color:blue" align="center">Pandas Time Series Analysis Tutorial: to_datetime</h1>

```python
# Parse a list of mixed date formats into a consistent datetime format using pandas' mixed parser.
import pandas as pd
dates = ['2017-01-05', 'Jan 5, 2017', '01/05/2017', '2017.01.05', '2017/01/05','20170105']
pd.to_datetime(dates, format='mixed')
```

```python
# Convert various date strings with times into datetime objects to demonstrate parsing of timestamps.
dt = ['2017-01-05 2:30:00 PM', 'Jan 5, 2017 14:30:00', '01/05/2016', '2017.01.05', '2017/01/05','20170105']
pd.to_datetime(dt)
```

<h3 style="color:purple">European style dates with day first</h3>

```python
# Show how pandas interprets a day-first European-style date string by default.
pd.to_datetime('30-12-2016')
```

```python
# Force pandas to interpret the date as day-first so the order of day and month is handled explicitly.
pd.to_datetime('5-1-2016', dayfirst=True)
```

<h3 style="color:purple">Custom date time format</h3>

```python
# Parse a custom-formatted date string by supplying the exact pattern used in the input.
pd.to_datetime('2017$01$05', format='%Y$%m$%d')
```

```python
# Demonstrate parsing of another custom delimiter format using a format string.
pd.to_datetime('2017#01#05', format='%Y#%m#%d')
```

<h3 style="color:purple">Handling invalid dates</h3>

```python
# Handle invalid date input by coercing bad values to NaT instead of raising an error.
pd.to_datetime(['2017-01-05', 'Jan 6, 2017', 'abc'], errors='coerce')
```

```python
# Show the same invalid-date example again to highlight how pandas converts problematic values to missing timestamps.
pd.to_datetime(['2017-01-05', 'Jan 6, 2017', 'abc'], errors='coerce')
```

<h3 style="color:purple">Epoch</h3>

**Epoch or Unix time means number of seconds that have passed since Jan 1, 1970 00:00:00 UTC time**

```python
# Convert a Unix timestamp in seconds into a human-readable datetime value.
current_epoch = 1501324478
pd.to_datetime(current_epoch, unit='s')
```

```python
# Convert the same timestamp from milliseconds into a datetime object to show epoch units.
pd.to_datetime(current_epoch*1000, unit='ms')
```

```python
# Create a datetime array from an epoch value and inspect the resulting pandas object.
t = pd.to_datetime([current_epoch], unit='s')
t
```

```python
# View the underlying integer representation of the datetime values stored in the pandas object.
t.view('int64')
```
