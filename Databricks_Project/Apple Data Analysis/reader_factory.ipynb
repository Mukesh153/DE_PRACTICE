```python
class Datasource :
    '''
    Abstract Class
    '''

    def __init__(self,path) :
        self.path = path
    
    def get_data_frame(self) :
        '''
        Abstract Method, function will be dfeined in child class
        '''
        raise ValueError("Not Implemented")

     
class Json(Datasource) :
    '''
    Class to read json file
    '''
    def get_data_frame(self) :
        return spark.read.json(self.path)
    
class CSV(Datasource) :
    '''
    Class to read csv file
    '''
    def get_data_frame(self) :
        return spark.read.csv(self.path,header=True)
    
class Parquet(Datasource) :
    '''
    Class to read parquet file
    '''
    def get_data_frame(self) :
        return spark.read.parquet(self.path)
    
class Delta(Datasource) :
    '''
    Class to read delta file
    '''
    def get_data_frame(self) :
        return spark.read.format("delta").load(self.path)

def get_data_frame(path) :
    '''
    Function to get data frame based on file type
    '''
    file_type = path.split('.')[-1]
    if file_type == 'json' :
        return Json(path).get_data_frame()
    elif file_type == 'csv' :
        return CSV(path).get_data_frame()
    elif file_type == 'parquet' :
        return Parquet(path).get_data_frame()
    elif file_type == 'delta' :
        return Delta(path).get_data_frame()
    else :
        raise ValueError("File type not supported")     
```
