```python
import pandas as pd

sf_restaurant_health_violations = spark.table("sf_restaurant_health_violations").toPandas()

df = sf_restaurant_health_violations[
    (sf_restaurant_health_violations["business_name"] == "Roxanne Cafe")
    & (sf_restaurant_health_violations["violation_id"].notna())
].copy()

df["inspection_date"] = pd.to_datetime(df["inspection_date"], errors="coerce")

sf_restaurant_health_violations_df = (
    df.dropna(subset=["inspection_date"])
      .assign(inspection_year=lambda x: x["inspection_date"].dt.year)
      .groupby("inspection_year")
      .size()
      .reset_index(name="n_violations")
)
```

```python
from pyspark.sql import functions as F
from pyspark.sql.types import *

df_sf_restaurant_health_violations = spark.table("sf_restaurant_health_violations")

sf_restaurant_health_violations_df = (
    df_sf_restaurant_health_violations
    .withColumn("inspection_date", F.to_date("inspection_date"))
    .filter(
        (F.col("business_name") == "Roxanne Cafe") &
        (F.col("violation_id").isNotNull())
    )
    .withColumn("inspection_year", F.year("inspection_date"))
    .groupBy("inspection_year")
    .count()
    .withColumnRenamed("count", "n_violations")
    .orderBy("inspection_year")
)

display(sf_restaurant_health_violations_df)
```
