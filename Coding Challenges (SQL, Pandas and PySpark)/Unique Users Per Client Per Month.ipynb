```python
import pandas as pd

df_spark = spark.table("fact_events")

df_pandas = df_spark.toPandas()

df_pandas['month'] = pd.to_datetime(df_pandas['time_id']).dt.month

df_agg = df_pandas.groupby(['client_id','month']).agg({'user_id':'nunique'}).reset_index()
df_agg.columns = ['client_id','month','users_num']

display(df_agg)
```

```python
from pyspark.sql import functions as F
#This would be written in pyspark language

df_spark = spark.table("fact_events")

df_spark = df_spark.withColumn("month", F.month(("time_id")))

df_agg = df_spark.groupBy(["client_id","month"]).agg(F.countDistinct("user_id").alias("users_num"))

df_agg.select("client_id","month","users_num").orderBy("client_id","month").show()

#This is the SQL query that was used to create the '_sqldf' Spark DataFrame
#SELECT client_id, month, COUNT(DISTINCT user_id) AS users_num FROM default.fact_events GROUP BY client_id, month ORDER BY client_id,
```
