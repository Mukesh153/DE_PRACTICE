```python
%sql
-- Calculate the ratio of employees with NULL phone numbers
select 
Cast((Sum (case when phone_number is null then 1 else 0 end)) * 1.0/ (count(*)) as decimal(10,2)) as null_phone_ratio from techcorp_workforce;
```

```python
import pandas as pd

techcorp_workforce_df = spark.table("techcorp_workforce").toPandas()

null_count = techcorp_workforce_df["phone_number"].isna().sum()
total_rows = len(techcorp_workforce_df)

null_phone_ratio = (null_count/total_rows) * 1.0

result = pd.DataFrame({
    "null_phone_ratio": [round(null_phone_ratio, 2)]
})

display(result)

```

```python
from pyspark.sql import functions as F
from pyspark.sql.types import *

df_techcorp_workforce = spark.table("techcorp_workforce")
techcorp_workforce_df_null_phone_ratio = (
    df_techcorp_workforce
    .agg(
        (F.sum(F.when(F.col("phone_number").isNull(), 1).otherwise(0)) / F.count("*")).alias("null_phone_ratio")
    )
)
display(techcorp_workforce_df_null_phone_ratio)
```
