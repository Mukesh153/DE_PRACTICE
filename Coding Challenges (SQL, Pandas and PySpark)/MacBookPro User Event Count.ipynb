```python
import pandas as pd

df_Playbook_Events = spark.table("playbook_events")

df_Playbook_Events = df_Playbook_Events.toPandas()

df_playbook_events = df_Playbook_Events[df_Playbook_Events['device']=='macbook pro'] 

df_playbook_events = df_playbook_events.groupby("event_name").size().reset_index(name = "event_count")

df_playbook_events.display()
```

```python
from pyspark.sql import functions as F

playbook_events = spark.table("playbook_events")

df_playbook_events_spark = playbook_events.where(F.col("device")=="macbook pro")

df_playbook_events_spark = df_playbook_events_spark.groupBy("event_name").agg(F.count("event_name").alias("event_count"))

df_playbook_events_spark = df_playbook_events_spark.select("event_name","event_count").orderBy("event_count",ascending=False)

df_playbook_events_spark.show()
```
